```python
%sql
CREATE CATALOG IF NOT EXISTS gizmobox
  MANAGED LOCATION 'abfss://gizmobox@optronstore.dfs.core.windows.net'
    COMMENT 'This is the catalog for the GizmoBox Data Lakehouse' 
```

```python
%sql
select current_catalog();
    
SHOW SCHEMAS IN gizmobox;
    
USE CATALOG gizmobox;
```

```python
%sql

--DROP SCHEMA IF EXISTS landing;
CREATE SCHEMA IF NOT EXISTS landing
     MANAGED LOCATION 'abfss://gizmobox@optronstore.dfs.core.windows.net/landing' 
      COMMENT 'This is the landing zone for the GizmoBox Data Lakehouse'; 

CREATE SCHEMA IF NOT EXISTS bronze
     MANAGED LOCATION 'abfss://gizmobox@optronstore.dfs.core.windows.net/bronze' 
      COMMENT 'This is the bronze zone for the GizmoBox Data Lakehouse'; 

CREATE SCHEMA IF NOT EXISTS silver
     MANAGED LOCATION 'abfss://gizmobox@optronstore.dfs.core.windows.net/silver' 
      COMMENT 'This is the silver zone for the GizmoBox Data Lakehouse'; 

CREATE SCHEMA IF NOT EXISTS gold
     MANAGED LOCATION 'abfss://gizmobox@optronstore.dfs.core.windows.net/gold' 
      COMMENT 'This is the gold zone for the GizmoBox Data Lakehouse'; 
```
