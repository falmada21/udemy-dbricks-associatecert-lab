Extract Customer ID from the File_Path

```python
%sql

SELECT regexp_extract(path, '.*/([0-9]+)\\.png$', 1) AS customer_id,
       content as membership
FROM gizmobox.bronze.v_memberships;
```

```python
%sql
CREATE TABLE gizmobox.silver.memberships 
AS 
SELECT regexp_extract(path, '.*/([0-9]+)\\.png$', 1) AS customer_id,
       content as membership
FROM gizmobox.bronze.v_memberships;
```
