### Transform Addresses data

```python
%sql

SELECT * 
FROM 
  (SELECT * FROM gizmobox.bronze.v_addresses)
PIVOT(MAX(address_line_1) AS address_1,
      MAX(city) as city,
      MAX(state) as state,
      MAX(postcode) as postcode
      FOR address_type IN ('shipping', 'billing')  
);
```

```python
%sql
CREATE TABLE gizmobox.silver.addresses 
AS
SELECT * 
FROM 
  (SELECT * FROM gizmobox.bronze.v_addresses)
PIVOT(MAX(address_line_1) AS address_1,
      MAX(city) as city,
      MAX(state) as state,
      MAX(postcode) as postcode
      FOR address_type IN ('shipping', 'billing')  
);
```
