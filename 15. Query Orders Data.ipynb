Extract Top Level Object Values

```python
%sql

SELECT value:order_id AS order_id, -- Column Values
       value
FROM gizmobox.bronze.v_orders;
```

2. Extract Array Elements

```python
%sql

SELECT value:items[0] AS item_1,
       value:items[1] AS item_2,
       value:items[2] AS item_3
FROM gizmobox.bronze.v_orders;
```
