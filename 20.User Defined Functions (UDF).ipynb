
#User Defined Functions (UDF)
User Defined Functions in Spark are custom functions created by users to extend the capabilities of Spark SQL
UDFs allow us to perform calculations or transformations to apply business logic
Define the function once and use it accross multiple queries. 
SQL UDFs are recommended over python UDFs 

CREATE OR REPLACE FUNCTION catalog_name.schema.udf_name(param_name)
RETURNS return_type
RETURNS expression;


```python
%sql
CREATE OR REPLACE FUNCTION gizmobox.default.get_fullname(firstname STRING, surname STRING)
RETURNS STRING
RETURN concat(initcap(firstname), ' ',initcap(surname));
```

```python
%sql
SELECT gizmobox.default.get_fullname('john', 'doe');
```

```python
%sql DESC FUNCTION EXTENDED gizmobox.default.get_fullname
```

```python

%sql
CREATE OR REPLACE FUNCTION gizmobox.default.get_payment_status(payment_status INT)
RETURNS STRING
RETURN CASE payment_status
WHEN 1 THEN 'Success'
WHEN 2 THEN 'Pending'
WHEN 3 THEN 'Cancelled'
WHEN 4 THEN 'Failed'
END;
    
```
