```python
%sql

CREATE OR REPLACE TEMPORARY VIEW order_items AS 
SELECT * FROM VALUES
(1,array('smartphone','laptop', 'monitor')),
(2, array('headphones','mouse','keyboard')),
(3, array('shirts','jeans','sneakers')),
(4, array('dress','blouse','hat')),
(5, array('socks','t-shirt','jacket'))
AS orders(order_id, item);

```

```python
%sql

SELECT * FROM order_items;

```

```python
%sql
SELECT order_id, 
       TRANSFORM(item, x -> UPPER(x)) AS upper_items 
FROM order_items;
```

```python
%sql
 
SELECT order_id, 
       FILTER(item, x -> x LIKE '%ouse%' ) AS smart_items 
FROM order_items;
```

```python
%sql
 -- Check weather exists

SELECT order_id, 
       EXISTS(array_filter(item.name, x -> x LIKE 'mouse')) AS smart_items 
FROM order_items;
```

```python
%sql

-- ARRAY with more than one subject
CREATE OR REPLACE TEMP VIEW order_items AS 
SELECT * FROM VALUES
      (1,array(
        named_struct('name','smartphone','price', 699),
        named_struct('name','laptop','price', 1199),
        named_struct('name','monitor','price', 1299)
      )),
      (2, array(
        named_struct('name','headphones','price', 199),
        named_struct('name','mouse','price', 99),
        named_struct('name','keyboard','price', 149)
      )),
      (3, array(
        named_struct('name','keyboard','price', 149),
        named_struct('name','mouse','price', 59)
        ))
      AS orders(order_id, item);
```

```python
%sql
SELECT * FROM order_items;
```

1. Convert all the item names to be UPPERCASE & Add 10% TAX to each item (TRANSFORM Function)

```python
%sql
SELECT 
order_id,
TRANSFORM(item, x -> named_struct(
                                      'name', UPPER(x.name),
                                      'price', x.price * 1.1)
    )

    FROM order_items
```
